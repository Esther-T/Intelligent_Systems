In this guide, we will be exploring the different frameworks used to build AI Agents (this content is derived from Huggingface Agents course)

The following are different Agentic Framework Units:

1. smolagents
- Agents framework developed by Hugging Face

2. Llama-Index
- End-to-end tooling to ship a context-augmented AI agent to production

3. LangGraph
- Agents allowing stateful orchestration of agents
